```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
from matplotlib.animation import FuncAnimation
import matplotlib.animation as animation
```

```python
x,y=make_regression(
    n_samples=100,
    n_features=1,
    n_informative=1,
    n_targets=1,
    noise=20,
    random_state=13
)
```

```python
plt.scatter(x,y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Regression Data')
plt.show()
```

```python
b=150
m=-127.82
lr=0.001
all_b=[]
all_m=[]
all_cost=[]
epochs=30
for i in range(epochs):
    slope_b=0
    slope_m=0
    cost=0
    for j in range(x.shape[0]):
        slope_b=slope_b -2*(y[j]-(m*x[j])-b)
        slope_m=slope_m-2*(y[j]-m*x[j])-b*x[j]
        cost=cost+(y[j]-m*x[j]-b)**2
    b=b-lr*slope_b
    m=m-lr*slope_m
    all_b.append(b)
    all_m.append(m)
    all_cost.append(cost)
```

```python
# MUST BE AT THE VERY TOP OF THE CELL TO ENABLE LIVE ANIMATION

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. Initialize the figure
fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(x, y)

# 2. Setup the x range and initial line
x_i = np.arange(-3, 3, 0.1)
line, = ax.plot(x_i, x_i * all_m[0] + all_b[0], 'r-', linewidth=2)

# 3. Update function for each epoch
def update(i):
    label = 'epoch {0}'.format(i + 1)
    line.set_ydata(x_i * all_m[i] + all_b[i])
    ax.set_xlabel(label)
    return line,

# 4. Run the live animation
anim = FuncAnimation(fig, update, repeat=True, frames=epochs, interval=500)
plt.show()
```
